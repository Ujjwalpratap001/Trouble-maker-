```python
import time
```

```python
size = 5_000_000
```

```python
my_list = list(range(size))
```

```python
start = time.time()
```

```python
python_doubled = [x * 2 for x in my_list]
```

```python
import numpy as np
import time

# Create a list of 5 million numbers
size = 5_000_000
my_list = list(range(size))
my_array = np.array(my_list)

# ----------------------------------------------------
# 1. SUPER SPEED (Comparing performance)
# ----------------------------------------------------

# Regular Python (Slow: must loop through every item)
start = time.time()
python_doubled = [x * 2 for x in my_list]
print("Python Time:", time.time() - start, "seconds")

# NumPy (Fast: doubles everything in one batch)
start = time.time()
numpy_doubled = my_array * 2
print("NumPy Time: ", time.time() - start, "seconds")
```
